## 第四部分：深入探討各種型別

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？
2. 對應的 CPython 物件結構名稱為何？
3. 它如何儲存資料？
4. 是否有固定大小限制？
5. 若資料規模持續增加，如何擴充儲存空間？

int
1.是
2.PyLongObject
3.以可變長度的 digit 陣列儲存整數
4.不是
5.重新配置（reallocate）更多記憶體

float
1.是
2.PyFloatObject
3.使用 IEEE 754 64-bit 浮點數
4.是
5.不能擴充

complex
1.是
2.PyComplexObject
3.由兩個 double 組成（實部+虛部）
4.是
5.無法擴充，兩部分皆固定 64 位元

bool
1.是
2.PyBoolObject
3.只有 True 與 False 兩個物件，本質上是 int 子類別
4.是
5.不需要擴充

set
1.是
2.PySetObject
3.使用雜湊表（Hash Table）
4.不是
5.負載因子過高時重新配置更大的雜湊表（resize）

list
1.是
2.PyListObject
3.動態陣列，儲存物件指標
4.不是
5.自動配置更大的陣列並複製資料

dict
1.是
2.PyDictObject
3.雜湊表儲存鍵值對
4.不是
5.自動擴大雜湊表並重新雜湊（rehash）
