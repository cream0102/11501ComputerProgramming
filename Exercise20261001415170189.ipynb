1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
將十進位小數轉換成二進位時，可以使用：小數部分持續乘以 2，取整數部分作為下一個二進位數字。
0.1 × 2 = 0.2  → 0
0.2 × 2 = 0.4  → 0
0.4 × 2 = 0.8  → 0
0.8 × 2 = 1.6  → 1
0.6 × 2 = 1.2  → 1
0.2 × 2 = 0.4  → 0
0.4 × 2 = 0.8  → 0
0.8 × 2 = 1.6  → 1
0.6 × 2 = 1.2  → 1
...
依序取出整數部分：000110011001100110011...
因此：0.1₁₀ = 0.00011001100110011001100110011...₂
也可以寫成：0.1₁₀ = 0.0001(1001)₂，其中：(1001)表示循環出現。

十進位的：0.1 = 1/10，而：10 = 2 × 5
在二進位系統中，只有分母可以表示成：2ⁿ的分數才能有限表示。
例如：
1/2 = 0.1₂
1/4 = 0.01₂
1/8 = 0.001₂
都能在有限位數內表示完成。
含有因數 5 無法寫成純粹的 1/(2ⁿ)，因此在二進位中必須無限循環。

現象：0.1 無法被精確表示
在十進位中，0.1看起來很簡單。但在二進位中卻是：0.000110011001100110011...無限循環。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
由於0.1是無限循環二進位小數，Python只能保留有限位數：0.100000000000000005551115123...實際儲存時會取最接近的數值。
預設使用：四捨六入五成雙
意思：
小於 5 捨去
大於 5 進位
剛好在中間時，取偶數
例如：
2.5 → 2
3.5 → 4
4.5 → 4
5.5 → 6
這樣做可以降低長期計算的累積誤差。

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```
print(0.1 + 0.2)會輸出：0.30000000000000004
原因：
實際存入記憶體的是：
0.1 ≈ 0.10000000000000000555...
0.2 ≈ 0.20000000000000001110...
因此：
0.1 + 0.2
≈ 0.3000000000000000444...
最後被顯示為：
0.30000000000000004

方法一：四捨五入
print(round(0.1+0.2, 1))

輸出：0.3

方法二：誤差比較
import math

print(math.isclose(0.1+0.2, 0.3))

輸出：True
因為比較時允許極小誤差。

方法三：Decimal
from decimal import Decimal

a = Decimal("0.1")
b = Decimal("0.2")

print(a+b)

輸出：0.3
這是精確十進位運算。
